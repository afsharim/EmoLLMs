```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load your dataset
data = pd.read_csv("/research/hal-afsharim/CSE842-Project/data/public_data/train/track_a/eng.csv")  # Replace with your actual path

# Function to convert a row into the desired format
def format_row(row):
    input_text = (
        "Task: Categorize the text's emotional tone as either 'neutral or no emotion' or "
        "identify the presence of one or more of the given emotions (anger, fear, joy, "
        "sadness, surprise).\n"
        f"Text: {row['text']}\nThis text contains emotions:"
    )
    emotions = []
    for emotion in ['Anger', 'Fear', 'Joy', 'Sadness', 'Surprise']:
        if row[emotion] == 1:
            emotions.append(emotion.lower())
    output_text = ', '.join(emotions) if emotions else 'neutral or no emotion'
    return input_text, output_text

# Apply the function to format the data
formatted_data = data.apply(format_row, axis=1)

# Convert to a DataFrame
formatted_df = pd.DataFrame(formatted_data.tolist(), columns=['input_text', 'output_text'])

# Split into train, dev, and test sets
train, temp = train_test_split(formatted_df, test_size=0.3, random_state=42)
dev, test = train_test_split(temp, test_size=0.5, random_state=42)

# Save splits to JSON files
train.to_json("train.json", orient="records", lines=True)
dev.to_json("dev.json", orient="records", lines=True)
test.to_json("test.json", orient="records", lines=True)

print("Data has been reformatted, split, and saved into train_data.json, dev_data.json, and test_data.json.")
```
